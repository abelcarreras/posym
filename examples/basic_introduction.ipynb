Introduction to symmetry objects
===========================

In this notebook I will detail a basic usage of the symmetry objects defined in posym library

Import required modules

```python
from posym import PointGroup, SymmetryObject
```

PointGroup contains the information about the point group symmetry table.  
http://gernot-katzers-spice-pages.com/character_tables/index.html  
This is an example for C2h group

Get point group information
---------------------------------

```python
pg = PointGroup(group='C2h')
print('\nPoint group table')
print(pg)
```

This object implements different method to access to the raw data

```python
print('Number of symmetry elements: ', pg.n_sym_elements)
print('Irreducible representation labels: ', pg.ir_labels)
print('Symmetry operation labels: ', pg.op_labels)
print('Group order (h): ', pg.order)
print('is abelian?: ', pg.is_abelian)

print('\nSymmetry operations: ')
for operation in pg.operations:
    print(' ', operation)
    # print(operation.matrix_representation)
```

Create symmetry objects
-----------------------------

Symmetry objects are abstract representation of the symmetry of a particular object.  
For starting let's define a couple of generic symmetry objects

```python
ag = SymmetryObject(group='C2h', rep='Ag')
bu = SymmetryObject(group='C2h', rep='Bu')
```

we can print its string representation:

```python
print(ag)
```

get the point group to which this representation belong

```python
pg = ag.get_point_group()
print(pg)
```

Get its representation in symmetry operations and irreducible representation

```python
ag.get_op_representation()
```

```python
ag.get_ir_representation()
```

Symmetry operations
-------------------------
Symmetry objects have build-in defined operations using usual mathematical operators:  
- Direct product :  A * B
- Direct sum:  A + B

```python
ag + bu
```

```python
ag * bu
```

The result of these operations are another symmetry object that can be stored and manipulates like a usual variable. These operation can be combined following the usual operator order rules

```python
sym_obj = (bu + bu) * ag + ag
print(sym_obj)
```

Try out by yourself with a more complicated group:

```python
pg = PointGroup(group='Td')
print(pg)

a1 = SymmetryObject(group='Td', rep='A1')
a2 = SymmetryObject(group='Td', rep='A2')
e = SymmetryObject(group='Td', rep='E')
t1 = SymmetryObject(group='Td', rep='T1')
```
