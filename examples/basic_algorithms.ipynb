Usage of algorithms for symmetry analysis
=========================================

In order to determine the geometric symmetry, posym implements diferent algorithms to compute the permutation of atoms that minimize the CSM

Import required modules

```python
from posym import SymmetryMolecule
from posym.tools import get_basis_set_pyscf, build_density, build_orbital
from posym.algebra import norm
from posym.config import Configuration
import numpy as np
import time
import matplotlib.pyplot as plt
```

helper function for timing

```python
class TimeIt:
    def __init__(self):
        self._time_ini = 0
        self._time_fin = 0

    def __enter__(self):
        # Record the start time
        self._time_ini = time.time()
        return self  # So that `as f` receives this object

    def __exit__(self, exc_type, exc_value, traceback):
        # Record the end time
        self._time_fin = time.time()

    @property
    def elapsed(self):
        # Return the elapsed time
        return self._time_fin - self._time_ini
```

Algorithms to compute CSM
-------------------------

Define molecule (distorted ethane)

```python
coordinates = [[-0.7560,    0.0000,    0.0000],
               [ 0.7560,    0.0000,    0.0000],
               [-4.1404,    0.6586,    0.7845],
               [-2.1404,    0.3501,   -0.9626],
               [-1.1405,   -1.0087,    0.1781],
               [ 1.1404,   -0.3501,    0.9626],
               [ 2.1405,    1.0087,   -0.1781],
               [ 4.1404,   -0.6586,   -0.7845]]

symbols = ['C','C','H','H','H','H','H','H']
```

**Hungarian algorithm** : very fast, coherence only garanteed if close to exact symmetry (lower bound for the real CSM)

```python
Configuration().algorithm = 'hungarian'
Configuration().label_tolerance = 1.0

with TimeIt() as f:
  mb = SymmetryMolecule('C2h', coordinates, symbols, orientation_angles=[-90, 0.0, 109.987640])
  print('CSM: ', mb.measure)
  print('Coor measure: ', mb, '(', norm(mb), ')')
  print(mb._symbols_restricted)
print('time:', f.elapsed)
```

**Exact algorithm** : very slow but exact for any geometry and any group

```python

Configuration().algorithm = 'exact'
Configuration().label_tolerance = 1.0

with TimeIt() as f:
  mb = SymmetryMolecule('C2h', coordinates, symbols, orientation_angles=[-90, 0.0, 109.987640])
  print('CSM: ', mb.measure)
  print('Coor measure: ', mb, '(', norm(mb), ')')
  print(mb._symbols_restricted)
print('time:', f.elapsed)
```

**Exact algorithm with tolerance label**: faster as tolerance get close to zero, will yield overstimated CSM if tolerance label is too low (upper bound for the CSM)

```python
Configuration().algorithm = 'exact'
Configuration().label_tolerance = 0.2

with TimeIt() as f:
  mb = SymmetryMolecule('C2h', coordinates, symbols, orientation_angles=[-90, 0.0, 109.987640])
  print('CSM: ', mb.measure)
  print('Coor measure: ', mb, '(', norm(mb), ')')
  print(mb._symbols_restricted)
print('time:', f.elapsed)
```
